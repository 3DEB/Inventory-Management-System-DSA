# 9. Data Structures Used

Data Structures are fundamental to the design of the Inventory Management System because they determine how product information and inventory operations are stored, accessed, and managed.

The following data structures are proposed for the system:

### 9.1 Dictionary

A dictionary will be used as the primary structure for storing product records. Each product will be associated with a unique Product ID, which can be used as the key for efficient retrieval and updating of product information.

**Purpose:**

* Store product records using unique Product IDs.
* Provide fast average-case lookup and update operations.
* Maintain a direct relationship between a product identifier and its details.

### 9.2 List

Lists will be used where an ordered collection of products or records is required.

**Purpose:**

* Maintain collections of inventory records.
* Support sequential processing of products.
* Provide input data for sorting and analysis operations.

### 9.3 Queue

A Queue will be used to manage pending reorder requests. It follows the First-In, First-Out (FIFO) principle, meaning that the earliest pending request can be processed first.

**Purpose:**

* Maintain pending reorder requests.
* Process requests in their order of arrival.
* Demonstrate the practical application of the FIFO principle.

### 9.4 Stack

A Stack will be used to maintain the history of inventory operations. It follows the Last-In, First-Out (LIFO) principle, allowing the most recent operation to be accessed first.

**Purpose:**

* Store recent inventory operations.
* Maintain operation history.
* Support future implementation of undo functionality.

### 9.5 Data Structure Selection

Each data structure is selected according to the requirements of the operation it supports.

| Data Structure | Primary Purpose                      | Principle  |
| -------------- | ------------------------------------ | ---------- |
| Dictionary     | Product lookup and record management | Key-Value  |
| List           | Ordered product collections          | Sequential |
| Queue          | Reorder request management           | FIFO       |
| Stack          | Operation history                    | LIFO       |

The combination of these structures allows the system to demonstrate multiple fundamental Data Structures and their practical application in an inventory management environment.


# 10. Algorithms Used

Algorithms provide the logical procedures required to perform operations efficiently within the Inventory Management System. Appropriate searching, sorting, and inventory-processing techniques will be implemented according to the requirements of the system.

### 10.1 Linear Search

Linear Search examines inventory records sequentially until the required product is found or all records have been checked.

**Purpose:**

* Search through product records sequentially.
* Handle searches when data is not sorted.
* Provide a simple baseline for search performance comparison.

**Time Complexity:**

* Best Case: O(1)
* Average Case: O(n)
* Worst Case: O(n)

### 10.2 Binary Search

Binary Search will be used for searching product records when the data is maintained in an appropriate sorted order.

The algorithm repeatedly divides the search range into two parts and determines which half may contain the required product.

**Purpose:**

* Perform efficient searching on sorted data.
* Demonstrate the performance difference between linear and binary search.

**Time Complexity:**

* Best Case: O(1)
* Average Case: O(log n)
* Worst Case: O(log n)

### 10.3 Merge Sort

Merge Sort will be used to organize inventory records according to selected attributes such as product name, price, or quantity.

The algorithm follows the divide-and-conquer approach by dividing the data into smaller portions, sorting them, and then merging the sorted portions.

**Purpose:**

* Sort inventory records efficiently.
* Demonstrate the divide-and-conquer technique.
* Provide predictable sorting performance.

**Time Complexity:**

* Best Case: O(n log n)
* Average Case: O(n log n)
* Worst Case: O(n log n)

### 10.4 Inventory Processing Algorithms

In addition to searching and sorting, the system will use logical procedures for inventory operations such as:

* Adding a new product.
* Removing an existing product.
* Updating product information.
* Increasing stock through stock-in operations.
* Decreasing stock through stock-out operations.
* Identifying products below their reorder level.
* Adding and processing reorder requests.
* Recording inventory operations.

### 10.5 Algorithm Selection

The algorithms are selected according to the requirements of individual operations rather than using a single algorithm throughout the system.

| Algorithm     | Application               | Main Advantage                     |
| ------------- | ------------------------- | ---------------------------------- |
| Linear Search | General product searching | Works without sorted data          |
| Binary Search | Searching sorted records  | O(log n) search time               |
| Merge Sort    | Inventory sorting         | O(n log n) predictable performance |

The implementation and complexity analysis sections will demonstrate how these algorithms operate within the Inventory Management System.

### 19.5 Algorithm for Linear Search

**Purpose:**
To search for a product sequentially in the inventory.

**Steps:**

1. Start from the first product.
2. Compare the search key with the current product.
3. If a match is found, return the product.
4. Otherwise, move to the next product.
5. Continue until the product is found or all records are checked.
6. If no match is found, display "Product Not Found".
7. Stop.

**Pseudocode:**

```text
LINEAR_SEARCH(Key)

    FOR each Product in Inventory

        IF Product matches Key THEN
            Return Product
        END IF

    END FOR

    Return "Product Not Found"

END
```

**Time Complexity:**

* Best Case: O(1)
* Average Case: O(n)
* Worst Case: O(n)

### 19.6 Algorithm for Binary Search

**Purpose:**
To efficiently search for a product in sorted inventory records.

**Condition:**
The inventory must be sorted according to the search key before applying Binary Search.

**Steps:**

1. Set the lower boundary to the first record.
2. Set the upper boundary to the last record.
3. Calculate the middle position.
4. Compare the middle record with the search key.
5. If the key matches, return the product.
6. If the key is greater, search the right half.
7. If the key is smaller, search the left half.
8. Repeat until the product is found or the search range becomes empty.
9. Display "Product Not Found" if no match exists.

**Pseudocode:**

```text
BINARY_SEARCH(Inventory, Key)

    Low = 0
    High = Length(Inventory) - 1

    WHILE Low <= High

        Mid = (Low + High) // 2

        IF Inventory[Mid] matches Key THEN
            Return Inventory[Mid]

        ELSE IF Inventory[Mid] is less than Key THEN
            Low = Mid + 1

        ELSE
            High = Mid - 1
        END IF

    END WHILE

    Return "Product Not Found"

END
```

**Time Complexity:**

* Best Case: O(1)
* Average Case: O(log n)
* Worst Case: O(log n)

---

### 19.7 Algorithm for Merge Sort

**Purpose:**
To arrange inventory records according to Product ID, Product Name, Price, or Quantity.

Merge Sort uses the **divide-and-conquer** approach.

**Steps:**

1. Start with the inventory list.
2. Check whether the list contains one or zero elements.
3. If yes, return the list.
4. Divide the list into two halves.
5. Apply Merge Sort to the left half.
6. Apply Merge Sort to the right half.
7. Merge the two sorted halves.
8. Return the sorted inventory.

**Pseudocode:**

```text
MERGE_SORT(Inventory)

    IF Length(Inventory) <= 1 THEN
        Return Inventory
    END IF

    Find Middle Position

    Left = First Half of Inventory
    Right = Second Half of Inventory

    Left = MERGE_SORT(Left)
    Right = MERGE_SORT(Right)

    Return MERGE(Left, Right)

END
```

**Merge Pseudocode:**

```text
MERGE(Left, Right)

    Create Empty Result

    WHILE Left is not empty AND Right is not empty

        IF First element of Left <= First element of Right THEN
            Add First element of Left to Result
        ELSE
            Add First element of Right to Result
        END IF

    END WHILE

    Add remaining elements of Left to Result
    Add remaining elements of Right to Result

    Return Result

END
```

**Time Complexity:**

* Best Case: O(n log n)
* Average Case: O(n log n)
* Worst Case: O(n log n)

**Space Complexity:** O(n)

### 19.9 Algorithm for Stock-Out

**Purpose:**
To decrease product quantity when products are sold, issued, or removed from inventory.

**Steps:**

1. Enter Product ID.
2. Search for the product.
3. If the product does not exist, display an error.
4. Enter the Stock-Out quantity.
5. Check whether sufficient stock is available.
6. If stock is insufficient, reject the transaction.
7. Otherwise, subtract the quantity from available stock.
8. Save the updated inventory.
9. Display the remaining quantity.
10. Stop.

**Pseudocode:**

```text
STOCK_OUT(Product_ID, Quantity)

    Search for Product_ID

    IF Product does not exist THEN
        Display "Product Not Found"
        Return
    END IF

    IF Quantity <= 0 THEN
        Display "Invalid Quantity"
        Return
    END IF

    IF Quantity > Product.Quantity THEN
        Display "Insufficient Stock"
        Return
    END IF

    Product.Quantity =
        Product.Quantity - Quantity

    Save Inventory

    Display "Stock-Out Successful"
    Display Remaining Quantity

END
```

5. Display the low-stock product.
6. Continue until all products are checked.
7. Stop.

**Pseudocode:**

```text
CHECK_LOW_STOCK()

    Low_Stock_Found = FALSE

    FOR each Product in Inventory

        IF Product.Quantity <= Product.Reorder_Level THEN

            Display Product
            Low_Stock_Found = TRUE

        END IF

    END FOR

    IF Low_Stock_Found = FALSE THEN
        Display "No Low-Stock Products"
    END IF

END
```

### 21.7 Linear Search Implementation

Linear Search is implemented for searching inventory records sequentially.

The algorithm starts from the first product and compares the search key with each record until a match is found.

```text id="r0bqv0"
FOR each product in inventory:
    IF product ID matches search ID:
        Display product
        Stop search
```

If all records are checked without finding a match, the system displays:

```text
Product not found.
```

### 21.8 Binary Search Implementation

Binary Search is implemented for searching records that have already been sorted according to the selected search key.

The implementation uses:

* Low index.
* High index.
* Middle index.

At every iteration, approximately half of the remaining search area is eliminated.

This provides efficient searching with a worst-case time complexity of **O(log n)**.

### 21.9 Merge Sort Implementation

Merge Sort is implemented to arrange inventory records according to selected attributes such as:

* Product ID.
* Product Name.
* Price.
* Quantity.

The algorithm divides the inventory into smaller portions, recursively sorts them, and merges them into a final sorted list.

The implementation follows the divide-and-conquer approach.

**Time Complexity:** O(n log n)

**Space Complexity:** O(n)


### 21.19 Implementation Summary

The major implementation components of the project are:

| Component            | Implementation                         |
| -------------------- | -------------------------------------- |
| Programming Language | Python 3.x                             |
| Data Structure       | List and Dictionary                    |
| Searching            | Linear Search and Binary Search        |
| Sorting              | Merge Sort                             |
| Inventory Processing | Stock-In and Stock-Out                 |
| Low-Stock Detection  | Conditional checking and traversal     |
| Storage              | CSV / JSON / Text File                 |
| Interface            | Console-based menu                     |
| Validation           | Input and inventory validation         |
| Authentication       | Basic username and password validation |

The implementation combines these components into a structured DSA-based application. This demonstrates how theoretical concepts such as **data structures, searching, sorting, traversal, and algorithmic complexity** can be applied to solve a practical inventory management problem.


# 22. Complexity Analysis

Complexity analysis is an important part of the **Inventory Management System** because it measures the efficiency of the Data Structures and Algorithms used in the project. It helps determine how the execution time and memory requirements of the system change as the number of inventory records increases.

Let **n** represent the total number of product records stored in the inventory.

The two main types of complexity considered in this project are:

* **Time Complexity:** Measures how the execution time of an operation increases with the input size.
* **Space Complexity:** Measures how much additional memory an algorithm requires as the input size increases.

### 22.1 Linear Search

Linear Search checks inventory records one by one until the required product is found.

**Time Complexity:**

* Best Case: **O(1)**
* Average Case: **O(n)**
* Worst Case: **O(n)**

**Space Complexity:**

* **O(1)** for an iterative implementation.

Linear Search is simple and can be applied directly to unsorted inventory records.

---

### 22.2 Binary Search

Binary Search operates on sorted inventory records and repeatedly divides the search range into two parts.

**Time Complexity:**

* Best Case: **O(1)**
* Average Case: **O(log n)**
* Worst Case: **O(log n)**

**Space Complexity:**

* **O(1)** for an iterative implementation.

Binary Search is more efficient than Linear Search for large sorted datasets.

---

### 22.3 Merge Sort

Merge Sort uses the divide-and-conquer technique to sort inventory records.

**Time Complexity:**

* Best Case: **O(n log n)**
* Average Case: **O(n log n)**
* Worst Case: **O(n log n)**

**Space Complexity:**

* **O(n)**

Merge Sort provides predictable performance and is suitable for sorting a large number of inventory records.

---

### 22.4 Add Product

Adding a product to the inventory list generally requires constant time after validation.

**Insertion Time Complexity:**

* **O(1)** average case

However, checking whether the Product ID already exists may require searching through existing records.

**Duplicate Checking Complexity:**

* **O(n)** using Linear Search.

Therefore, the overall operation can be considered **O(n)** when duplicate Product ID validation is included.

---

### 22.5 Delete Product

To delete a product, the system first searches for the Product ID.

Using Linear Search:

**Time Complexity:**

* Best Case: O(1)
* Average Case: O(n)
* Worst Case: O(n)

Removing an element from a Python list may also require shifting subsequent elements.

Therefore, the overall deletion operation is generally:

**Time Complexity:** O(n)

**Space Complexity:** O(1) additional space.

---

### 22.6 Update Product

The Update Product operation first searches for the required Product ID and then modifies its information.

Using Linear Search:

**Time Complexity:**

* Best Case: O(1)
* Average Case: O(n)
* Worst Case: O(n)

The actual modification of the product record requires constant time after the record is located.

**Overall Time Complexity:** O(n)

---

### 22.7 Stock-In

The Stock-In operation requires the system to locate the product before increasing its quantity.

Using Linear Search:

**Search Complexity:** O(n)

Updating the quantity:

**Update Complexity:** O(1)

Therefore:

**Overall Time Complexity:** O(n)

If a suitable direct-access structure is used for product lookup, the update portion can be performed more efficiently.

---

### 22.8 Stock-Out

The Stock-Out operation first searches for the product and then verifies whether sufficient stock is available.

The process involves:

1. Searching for the product.
2. Checking available quantity.
3. Deducting the requested quantity.
4. Saving the updated record.

Using Linear Search:

**Overall Time Complexity:** O(n)

The actual quantity update takes:

**O(1)**

---

### 22.9 Low-Stock Detection

The system checks every product and compares its quantity with the reorder level.

```text id="v9t1w0"
IF Quantity <= Reorder_Level
    Mark Product as Low Stock
```

Since every inventory record must be checked:

**Time Complexity:** O(n)

**Space Complexity:** O(1) additional space if results are displayed directly.

---

### 22.10 Display Inventory

The system traverses every product record and displays its information.

For **n** products:

**Time Complexity:** O(n)

**Space Complexity:** O(1) additional space, excluding the memory already occupied by the inventory records.

---

### 22.11 Sorting Inventory

When the user requests inventory sorting, Merge Sort is applied to the records.

**Time Complexity:** O(n log n)

**Space Complexity:** O(n)

The complexity remains O(n log n) for the best, average, and worst cases.

---

### 22.12 File Reading and Writing

File operations depend on the number of inventory records being processed.

If the complete inventory file contains **n** records:

**Reading all records:**

* Time Complexity: O(n)

**Writing all records:**

* Time Complexity: O(n)

The exact performance may also depend on the file format and storage device.

# 26. Limitations

Although the **Inventory Management System** provides an organized and efficient approach to managing inventory records, the current version has certain limitations. These limitations are mainly related to the project scope, storage method, user interface, security, and scalability of the current implementation.

### 26.1 File-Based Storage

The current system uses file-based storage such as CSV or JSON files instead of a dedicated database.

This may create limitations when:

* The number of records becomes very large.
* Multiple users access the system simultaneously.
* Frequent data updates are required.
* Advanced database queries are needed.

### 26.2 Single-User Environment

The current version is primarily designed for a single-user or administrator environment.

It does not provide complete support for multiple users working on the same inventory simultaneously.

### 26.3 Basic Authentication

The Login Module provides basic username and password authentication.

The current implementation does not include advanced security features such as:

* Password encryption.
* Multi-factor authentication.
* Role-based access control.
* Account recovery.
* Login activity monitoring.

### 26.4 Console-Based Interface

The current system uses a console-based interface.

Although the menu-driven interface is simple and suitable for demonstrating DSA concepts, it does not provide the visual experience of a modern graphical or web-based application.

### 26.5 Limited Reporting

The current reporting functionality focuses mainly on basic inventory information such as:

* Available products.
* Low-stock products.
* Product quantities.
* Inventory records.

Advanced analytical reports such as sales trends, profit analysis, supplier performance, and demand forecasting are not included in the current version.

### 26.6 No Barcode or QR Code Integration

The system does not currently support barcode or QR code scanning.

Product information must be entered manually using the keyboard.

### 26.7 Limited Supplier Management

Supplier information is stored as part of the product record. A separate supplier management system with detailed supplier profiles, purchase orders, delivery tracking, and supplier performance analysis is not included.

### 26.8 No Real-Time Synchronization

The current application does not provide real-time synchronization between multiple devices or locations.

Changes made on one computer are not automatically reflected on another system.

### 26.9 Limited Scalability

The current file-based approach is suitable for a small or medium-sized inventory but may become less efficient as the number of records grows significantly.

For large-scale applications, a database management system would be more appropriate.

### 26.10 Limited Backup and Recovery

The current version does not provide an advanced automated backup and disaster recovery mechanism.

If inventory files are accidentally deleted or corrupted, recovery may require a manually maintained backup.

### 26.11 Limited Error Handling

Basic input validation and error handling are implemented, but the system may require additional exception handling for advanced situations such as:

* Corrupted inventory files.
* Unexpected file access errors.
* Invalid file formats.
* Storage failures.
* Simultaneous file modifications.

### 26.12 No Advanced Data Analytics

The current project focuses mainly on inventory management and DSA implementation.

It does not currently include advanced analytics such as:

* Demand prediction.
* Sales forecasting.
* Stock trend analysis.
* Product recommendation.
* Machine learning-based inventory prediction.

### 26.13 Manual Data Entry

Product details, Stock-In quantities, and Stock-Out quantities are entered manually.

Manual data entry can result in incorrect information if the user enters invalid or inaccurate values.

### 26.14 Limited Transaction Management

The current implementation records basic inventory operations but does not provide a complete enterprise-level transaction management system with detailed invoices, purchase orders, sales orders, payment information, and audit trails.
